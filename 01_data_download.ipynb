# Download Annual Peak Flow Data
## Overview
This notebook downloads annual peak-flow data from the United States Geological Survey National Water Information System for the selected streamflow-gauging station. The USGS station number is used to generate the annual peak-flow data URL, and the original dataset is downloaded in RDB format. The raw file is then stored in the project’s raw-data directory without any cleaning, transformation, or unit conversion. Preserving the original dataset provides a reliable record of the source data and allows the data-processing procedure to be reproduced.


```python
# Import Required Libraries
from pathlib import Path
from urllib.request import urlopen
```

## 1. Define the USGS station and download URL

```python
# Change this station number for a different study area
SITE_NO = "06214500"  

# USGS annual peak-flow data URL
URL = (
    "https://nwis.waterdata.usgs.gov/nwis/peak?"
    f"site_no={SITE_NO}&agency_cd=USGS&format=rdb"
)

print("USGS station number:", SITE_NO)
print("Download URL:", URL)
```

## 2. Create the raw-data folder

```python
RAW_DIR = Path("data/raw")
RAW_DIR.mkdir(parents=True, exist_ok=True)

raw_file = RAW_DIR / f"usgs_{SITE_NO}_annual_peak_flow.rdb"

print("Raw file location:", raw_file)
```

## 3. Download and save the raw USGS file

```python
print("Downloading annual peak-flow data from USGS...")

with urlopen(URL) as response:
    data = response.read()

with open(raw_file, "wb") as file:
    file.write(data)

print("Download complete.")
print(f"Raw data saved to: {raw_file}")
```
