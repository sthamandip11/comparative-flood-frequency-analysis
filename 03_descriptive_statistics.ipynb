# Descriptive Statistical Analysis
## Overview
This notebook performs a descriptive statistical analysis of the annual maximum flood-flow dataset. The purpose is to understand the central tendency, variability, distribution shape, and general behavior of the flood series before fitting probability distributions. Statistical measures such as the mean, median, standard deviation, variance, coefficient of variation, skewness, and kurtosis are calculated. The annual peak-flow series is also examined using a time-series plot, histogram, and boxplot.

```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 1. Load Clean Dataset

```python
# Change this station number for a different study area
SITE_NO = "06214500"

# Load clean flood dataset
df = pd.read_csv(f"data/processed/usgs_{SITE_NO}_clean_flood_data.csv")

df.head(2)
```

```python
# check shape of dataset
df.shape
```

## 2. Calculate Descriptive Statistics

```python
# Select flood data column
flow = df["Peak_Flow"]

# Calculate Descriptive Statistics
mean_flow = flow.mean()
median_flow = flow.median()
std_flow = flow.std()
variance_flow = flow.var()
cv_flow = std_flow / mean_flow
skewness_flow = flow.skew()
kurtosis_flow = flow.kurtosis()
```

```python
# Create Statistical Summary Table
stats_summary = pd.DataFrame({
    "Statistic": [
        "Mean",
        "Median",
        "Standard Deviation",
        "Variance",
        "Coefficient of Variation",
        "Skewness",
        "Kurtosis"
    ],
    "Value": [
        mean_flow,
        median_flow,
        std_flow,
        variance_flow,
        cv_flow,
        skewness_flow,
        kurtosis_flow
    ]
})

# cleaner display
stats_summary["Value"] = stats_summary["Value"].round(3)
stats_summary
```

```python
# Basic Built-in Summary
df.describe()
```

## 3. Plot of Annual Peak Flow

### 3.1 Basic Plot

```python
plt.figure(figsize=(10, 5))

plt.plot(df["Year"], df["Peak_Flow"], marker="o", linestyle="-")

plt.xlabel("Year")
plt.ylabel("Annual Peak Flow (m³/s)")
plt.title("Annual Maximum Flood Series")
plt.grid(True)

plt.show()
```

### 3.2 Histogram

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Peak_Flow"], bins=10, edgecolor="black")

plt.xlabel("Annual Peak Flow (m³/s)")
plt.ylabel("Frequency")
plt.title("Histogram of Annual Peak Flow")
plt.grid(True)

plt.show()
```

### 3.3 Boxplot for Outlier Visualization

```python
plt.figure(figsize=(6, 5))

plt.boxplot(df["Peak_Flow"], vert=True)

plt.ylabel("Annual Peak Flow (m³/s)")
plt.title("Boxplot of Annual Peak Flow")
plt.grid(True)

plt.show()
```
